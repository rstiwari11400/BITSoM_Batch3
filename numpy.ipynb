```python
import numpy as np
```

## 1. Create a NumPy Array

A NumPy array is like a Python list, but it is faster for numerical work.

Here we store monthly sales, in rupees, for one store.


```python
monthly_sales = np.array([120000, 150000, 135000, 170000, 160000, 180000])
monthly_sales
```

## 2. Array Properties

These properties help us understand the structure of an array.


```python
print("Shape:", monthly_sales.shape)
print("Size:", monthly_sales.size)
print("Dimensions:", monthly_sales.ndim)
print("Data type:", monthly_sales.dtype)
```

## 3. Indexing

Indexing is used to select one value from an array. Python indexing starts from 0.


```python
print("First month sales:", monthly_sales[0])
print("Third month sales:", monthly_sales[2])
print("Last month sales:", monthly_sales[-1])
```

## 4. Slicing

Slicing is used to select a range of values.


```python
print("First three months:", monthly_sales[0:3])
print("From fourth month onwards:", monthly_sales[3:])
print("Alternate months:", monthly_sales[::2])
```

## 5. Vectorized Operations

NumPy can perform calculations on the whole array at once.


```python
# Add Rs. 10,000 to every month
monthly_sales + 10000
```

```python
# Convert rupees to lakhs
monthly_sales / 100000
```

```python
# Increase every month's sales by 10 percent
monthly_sales * 1.10
```

## 6. Aggregation Functions

Aggregation functions summarize data.


```python
print("Total sales:", monthly_sales.sum())
print("Average sales:", monthly_sales.mean())
print("Minimum sales:", monthly_sales.min())
print("Maximum sales:", monthly_sales.max())
```

## 7. Boolean Filtering

Boolean filtering selects values that match a condition.


```python
monthly_sales > 150000
```

```python
monthly_sales[monthly_sales > 150000]
```

## 8. 2D Arrays

A 2D array is like a table.

Here, rows are Indian cities and columns are months.


```python
cities = np.array(["Mumbai", "Delhi", "Bengaluru"])
months = np.array(["Jan", "Feb", "Mar", "Apr"])

city_sales = np.array([
    [120000, 150000, 135000, 170000],
    [100000, 125000, 140000, 155000],
    [130000, 160000, 150000, 190000]
])

city_sales
```

```python
print("Shape:", city_sales.shape)
print("Rows:", city_sales.shape[0])
print("Columns:", city_sales.shape[1])
```

## 9. Selecting Data from a 2D Array

Use row index and column index to select values from a 2D array.


```python
# Sales of Mumbai across all months
city_sales[0]
```

```python
# Sales of all cities in February
city_sales[:, 1]
```

```python
# Sales of Delhi in March
city_sales[1, 2]
```

## 10. Axis Logic

- `axis=0` works down the rows, giving column-wise results.
- `axis=1` works across the columns, giving row-wise results.


```python
# Total sales for each month
city_sales.sum(axis=0)
```

```python
# Total sales for each city
city_sales.sum(axis=1)
```

```python
# Average sales for each city
city_sales.mean(axis=1)
```

## 11. Reshaping

Reshaping changes the shape of an array without changing its values.


```python
numbers = np.arange(1, 13)
numbers
```

```python
numbers.reshape(3, 4)
```

```python
numbers.reshape(2, 6)
```
