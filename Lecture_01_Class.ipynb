## Boolean Operations

Boolean operations work with the values `True` and `False`.

### AND

The `and` operator returns `True` only when **both** conditions are `True`.

| A | B | A and B |
|---|---|---------|
| True | True | True |
| True | False | False |
| False | True | False |
| False | False | False |

### OR

The `or` operator returns `True` when **at least one** condition is `True`.

| A | B | A or B |
|---|---|--------|
| True | True | True |
| True | False | True |
| False | True | True |
| False | False | False |

### NOT

The `not` operator reverses a Boolean value.

| A | not A |
|---|-------|
| True | False |
| False | True |

# String slicing

```python
sentence = 'This is a short sentence about Python.'
print('Print First Character---------------------------')
print(sentence[0])

print('Print Last Character---------------------------')    
print(sentence[-1])

print('get first four characters---------------------------')
print(sentence[:4])

print('get characters from index 5 to 10---------------------------')
print(sentence[5:11])

print('get alternate characters---------------------------')
print(sentence[::2])

print('get characters in reverse order---------------------------')
print(sentence[::-1])

```
