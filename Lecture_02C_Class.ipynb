## 3. Common Errors and Exception Handling


### Basic try-except


```python
b=int(input())
A=100
try:
    print(A/b)
except Exception as e:
    print('Error:', e)
```

```python
try:
    result = 10 / 0
    print(result)
except ZeroDivisionError:
    print('Cannot divide by zero')
```

### IndentationError


```python
try:
    exec("if True:\nprint('Hello')")
except IndentationError:
    print('Indentation is missing')
```

### ZeroDivisionError


```python
try:
    result = 10 / 0
    print(result)
except ZeroDivisionError:
    print('Cannot divide by zero')
```

### ValueError


```python
try:
    age = int('twenty')
    print(age)
except ValueError:
    print('Please enter a valid number')
```

### IndexError


```python
fruits = ['Apple', 'Banana']

try:
    print(fruits[1])
except IndexError:
    print('This index does not exist')
```

### KeyError


```python
student = {'name': 'Ravi', 'age': 21}

try:
    print(student['city'])
except KeyError:
    print(None)
```

### TypeError


```python
try:
    total = 10 + '5'
    print(total)
except TypeError:
    print('Cannot add different data types directly')
```

### AttributeError


```python
name = None

try:
    print(name.upper())
except AttributeError:
    print('This object does not have that method')
```

### Multiple except blocks and generic Exception


```python
try:
    number = int('10')
    result = number / 0
    print(result)
except ValueError:
    print('Invalid number')
except ZeroDivisionError:
    print('Cannot divide by zero')
except Exception as error:
    print('Something else went wrong:', error)
```

### Custom exception: InvalidOrderError


```python
class InvalidOrderError(Exception):
    pass

def place_order(quantity):
    if quantity <= 0:
        raise InvalidOrderError('Quantity must be greater than zero')
    print('Order placed')

try:
    place_order(0)
except InvalidOrderError as error:
    print(error)
```
