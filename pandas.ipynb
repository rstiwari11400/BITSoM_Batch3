## 1. Create a DataFrame

A DataFrame is a table with rows and columns. Here is a small sales dataset for an Indian retail store.


```python
sales_data = {
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "customer": ["Aarav", "Diya", "Kabir", "Meera", "Ishaan", "Ananya", "Rohan", "Sara"],
    "city": ["Mumbai", "Delhi", "Bengaluru", "Chennai", "Mumbai", "Pune", "Delhi", "Hyderabad"],
    "product": ["Laptop", "Phone", "Headphones", "Tablet", "Phone", "Laptop", "Tablet", "Headphones"],
    "quantity": [1, 2, 3, 1, 1, 2, 2, 4],
    "price": [55000, 18000, 2500, 22000, 20000, 52000, 24000, 2200],
    "payment_mode": ["UPI", "Card", "UPI", "Cash", "UPI", "Net Banking", "Card", "UPI"]
}

sales = pd.DataFrame(sales_data)
sales
```

## 6. Add New Columns

We can calculate the total order value using quantity and price.


```python
sales["order_value"] = sales["quantity"] * sales["price"]
sales
```

```python
sales["gst_18_percent"] = sales["order_value"] * 0.18
sales["final_amount"] = sales["order_value"] + sales["gst_18_percent"]
sales
```

## 7. Sort Data

Sorting makes it easy to find the highest or lowest values.


```python
sales.sort_values(by="final_amount", ascending=False)
```

## 8. Group and Summarize Data

`groupby` is useful for creating summaries such as sales by city or sales by payment mode.


```python
sales.groupby("city")["final_amount"].sum()
```

```python
sales.groupby("payment_mode")["order_value"].mean()
```

```python
sales.groupby("product").agg(
    total_quantity=("quantity", "sum"),
    total_sales=("final_amount", "sum"),
    average_price=("price", "mean")
)
```

## 9. Missing Values

Real datasets often have missing values. Pandas represents missing values as `NaN`.


```python
student_data = {
    "student": ["Arjun", "Neha", "Farhan", "Priya", "Tanya"],
    "city": ["Jaipur", "Kolkata", "Lucknow", None, "Ahmedabad"],
    "maths": [88, 92, None, 75, 81],
    "english": [79, None, 84, 90, 76]
}

students = pd.DataFrame(student_data)
students
```

```python
students.isna()
```

```python
students.isna().sum()
```

```python
students_filled = students.copy()
students_filled["city"] = students_filled["city"].fillna("Unknown")
students_filled["maths"] = students_filled["maths"].fillna(students_filled["maths"].mean())
students_filled["english"] = students_filled["english"].fillna(students_filled["english"].mean())
students_filled
```

## 10. Merge DataFrames

Merging is like using VLOOKUP in Excel. We can combine data from two tables using a common column.


```python
customers = pd.DataFrame({
    "customer": ["Aarav", "Diya", "Kabir", "Meera", "Ishaan", "Ananya", "Rohan", "Sara"],
    "loyalty_level": ["Gold", "Silver", "Bronze", "Silver", "Gold", "Bronze", "Silver", "Gold"]
})

customers
```

```python
sales_with_loyalty = sales.merge(customers, on="customer", how="left")
sales_with_loyalty
```

## 12. Using `map()` and `apply()`

`map()` is commonly used on one column to replace or convert values.

`apply()` is used when we want to run a custom function on a column or across rows.


### `map()` Example

Let us map each payment mode to a simple payment category.


```python
payment_category = {
    "UPI": "Digital",
    "Card": "Digital",
    "Net Banking": "Digital",
    "Cash": "Offline"
}

sales["payment_category"] = sales["payment_mode"].map(payment_category)
sales[["customer", "payment_mode", "payment_category"]]
```

### `apply()` on One Column

Here we create a price band for each product price.


```python
def price_band(price):
    if price >= 50000:
        return "Premium"
    elif price >= 20000:
        return "Mid Range"
    else:
        return "Budget"

sales["price_band"] = sales["price"].apply(price_band)
sales[["product", "price", "price_band"]]
```

### `apply()` Across Rows

Use `axis=1` when the function needs values from multiple columns in the same row.


```python
def delivery_speed(row):
    if row["city"] in ["Mumbai", "Delhi", "Bengaluru"] and row["payment_category"] == "Digital":
        return "Fast Delivery"
    else:
        return "Standard Delivery"

sales["delivery_type"] = sales.apply(delivery_speed, axis=1)
sales[["customer", "city", "payment_category", "delivery_type"]]
```
