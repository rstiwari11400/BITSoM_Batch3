## 1. Lists — Basics


### Create a list


```python
my_list = [4, 'ArithmeticError', 3.14, True]
print(my_list)
print(type(my_list))
```

### Length and indexing


```python
print(len(my_list))
print(my_list[0], my_list[1], my_list[2], my_list[3])
```

```python
print(my_list[-1], my_list[-2])
```

### Slicing


```python
print(my_list[:2])
```

```python
print(my_list[::-1])
```

### Updating list items


```python
my_list[3] = False
print(my_list)
```

### Adding items to a list


```python
nested_list = [1, 2, [3, 4], 5]
nested_list.append(4)
print(nested_list)
```

```python
nested_list = [1, 2, [3, 4], 5]
nested_list.insert(1, 1.5)
print(nested_list)
```

```python
nested_list = [1, 2, [3, 4], 5]
nested_list.extend([6, 7])
print(nested_list)
```

### Removing items from a list


```python
nested_list = [1, 2, [3, 4], 5]
nested_list.remove(1)
print(nested_list)
```

```python
nested_list = [1, 2, [3, 4], 5]
popped = nested_list.pop()
print(nested_list, popped)
```

```python
nested_list = [1, 2, [3, 4], 5]
del nested_list[0]
print(nested_list)
```

### Membership checks


```python
my_numeric_list = [1, 2, 3, 4, 5]
print(2 in my_numeric_list)
print(7 in my_numeric_list)
```

### Numeric list functions


```python
print(sum(my_numeric_list))
print(min(my_numeric_list))
print(max(my_numeric_list))
```

### Looping through lists


```python
for number in my_numeric_list:
    print(number)
```

### Sorting lists


```python
fruits = ['Banana', 'Apple', 'Mango']
print(sorted(fruits))
print(fruits)
```

```python
fruits.sort()
print(fruits)
```

### Nested lists


```python
item_availability = [['Apple', 100, 20, 'Kg'], ['Banana', 50, 10, 'Kg']]
print(item_availability)
```

```python
print(item_availability[0])
print(item_availability[0][0])
print(item_availability[0][1])
```

### Bridge to dictionaries


```python
apple = ['Apple', 100, 20, 'Kg']
print(apple[0])  # item name
print(apple[1])  # price
print(apple[2])  # quantity
print(apple[3])  # unit
```

With lists, we must remember what each position means. Dictionaries make this clearer by storing values with names like `name`, `price`, `quantity`, and `unit`.
