# Lecture 03D: File Handling

This notebook demonstrates how to create folders, write files, append text, read files, and search for text inside a file.

## Setup

Create a folder named `MyFiles` in the current notebook working directory.

```python
from pathlib import Path
base = Path.cwd()
data_dir = base / "MyFiles"
data_dir.mkdir(exist_ok=True)

print("Working in:", data_dir)
```

```python
base
```

## 1. Write a New Text File

`w` means write mode. It creates a new file or overwrites an existing file.

```python
text_path = data_dir / "orders"

with open(text_path, "w", encoding="utf-8") as f1:
    f1.write("Order #1001: Burger x2, Fries1 x1\n")
    f1.write("Order #1002: Pizza x1, Coke2 x2\n")

print("Wrote:", text_path)
```

## 2. Append to the Same File

`a` means append mode. It adds new content at the end of the file.

```python
with open(text_path, "a", encoding="utf-8") as f:
    f.write("Order #1003: Sushi x3, Miso Soup x1\n")

print("Appended one line.")
```

## 3. Read the Whole File at Once

`r` means read mode.

```python
with open(text_path, "r", encoding="utf-8") as f:
    content = f.read()

print("--- File content (read) ---")
print(content)
```

## 4. Read Line by Line

Reading line by line is a good habit for larger files.

```python
print("--- File content (line by line) ---")

with open(text_path, "r", encoding="utf-8") as f:
    for number, line in enumerate(f, start=1):
        print(f"{number}:", line)
```

## 5. Read Only the First N Characters

```python
with open(text_path, "r", encoding="utf-8") as f:
    first_five = f.read(5)

print("First 5 chars:", repr(first_five))
```

## Find the Line Number of an Order

Search the file and return the line number where `Pizza` appears.

```python
line_number = None
search_text = "Pizza"
text_path = data_dir / "orders.txt"

with open(text_path, "r", encoding="utf-8") as f:
    for number, line in enumerate(f, start=1):
        if search_text in line:
            line_number = number
            break

print("Line number of 'Pizza' order:", line_number)
```
