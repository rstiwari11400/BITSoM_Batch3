# Dictionaries — Basics


### Why dictionaries?
Lists use positions. Dictionaries use names, called keys.


```python
apple = ['Apple', 100, 20, 'Kg']
print(apple[0])  # name
print(apple[1])  # price
```

```python
apple = {
    'name': 'Apple',
    'price': 100,
    'quantity': 20,
    'unit': 'Kg'
}
print(apple)
```

### Access values


```python
print(apple['name'])
print(apple['price'])
```

```python
print(apple.get('quantity'))
print(apple.get('discount'))
```

### Update values


```python
apple['price'] = 120
print(apple)
```

### Add new key-value pairs


```python
apple['available'] = True
print(apple)
```

### Remove key-value pairs


```python
removed_value = apple.pop('available')
print(apple)
print(removed_value)
```

### Check if a key exists


```python
print('price' in apple)
print('discount' in apple)
```

### Keys, values, and items


```python
print(apple.keys())
print(apple.values())
print(apple.items())
```

### Loop through a dictionary


```python
for key in apple:
    print(key, apple[key])
```

```python
for key, value in apple.items():
    print(key, value)
```

### List of dictionaries


```python
items = [
    {'name': 'Apple', 'price': 100, 'quantity': 20, 'unit': 'Kg'},
    {'name': 'Banana', 'price': 50, 'quantity': 10, 'unit': 'Kg'}
]
print(items)
```

```python
print(items[0]['name'])
print(items[1]['price'])
```

```python
for item in items:
    print(item['name'], item['price'])
```
